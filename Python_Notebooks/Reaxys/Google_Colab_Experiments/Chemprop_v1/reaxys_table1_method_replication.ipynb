# **Creating Chemprop Training Command**

```python
# creating a command from important arguments/parameters within the 'verbose.log' and 'args.json' files attached by the authors, as well as their SI,
# and printing it - '--gpu 0' ensures the utilised runtime is the GPU (A100) in this case
cmd = (
f'chemprop_train '
f'--data_path reaxys_train_filtered_no_reg_nums.csv '
f'--dataset_type regression '
f'--num_folds 5 '
f'--epochs 100 '
f'--number_of_molecules 2 '
f'--depth 3 '
f'--hidden_size 2300 '
f'--ffn_hidden_size 1900 '
f'--ffn_num_layers 3 '
f'--dropout 0.15 '
f'--aggregation mean '
f'--batch_size 50 '
f'--metric rmse '
f'--extra_metrics mae '
f'--seed 0 '
f'--gpu 0 '
f'--save_dir table1_reaxys_model_filtered'
        )

print(cmd)
```

```python
# executing Chemprop training command with the Reaxys training subset
! /usr/local/miniconda/bin/conda run -n chemprop15 \
bash -c "$cmd"
```
