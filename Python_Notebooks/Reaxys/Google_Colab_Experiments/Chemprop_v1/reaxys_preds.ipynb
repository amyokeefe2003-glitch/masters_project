# **Creating Chemprop Prediction Command**

```python
# creating a prediction command utilising the held-out Reaxys test subset and all 5 models produced during training on the Reaxys training subset - saving individual
# predictions by each model to the same CSV file as the averaged ensemble prediction for simple calculation of error metrics following execution

cmd_pred = (
    f'chemprop_predict '
    f'--test_path reaxys_test_filtered_no_reg_nums.csv '
    f'--checkpoint_dir /content/drive/MyDrive/chemprop_runs/table1_reaxys_model_filtered '
    f'--number_of_molecules 2 '
    f'--preds_path table1_reaxys_test_preds.csv '
    f'--individual_ensemble_predictions'
)
```

```python
# executing Chemprop prediction command with the Reaxys test subset
! /usr/local/miniconda/bin/conda run -n chemprop15 \
bash -c "$cmd_pred"
```
