# **Creating Chemprop Training Command**

```python
# creating a training command utilising the Reaxys training subset and the - '--gpu 0' ensures the utilised runtime is the GPU (A100) in this case
cmd = (
    f'chemprop train '
	  f'--config-path best_config.toml '
    f'--data-path reaxys_train_filtered_no_reg_nums.csv '
    f'--smiles-columns Chromophore Solvent '
    f'--target-columns \\"Absorption max (nm)\\" \\"log(e/mol-1 dm3 cm-1)\\" '
    f'--task-type regression '
    f'--num-replicates 5 '
    f'--data-seed 0 '
    f'--epochs 100 '
    f'--metric rmse '
    f'--output-dir /content/drive/MyDrive/chemprop_runs/modern_chemprop_reaxys'
    )

print(cmd)
```

```python
# executing Chemprop training command with the Reaxys training subset
! /usr/local/miniconda/bin/conda run -n newchemprop \
bash -c "$cmd"
```

# **Creating Chemprop Prediction Command**

```python
# creating a prediction command utilising the held-out Reaxys test subset and all 5 models produced during training on the Reaxys training subset - saving individual
# predictions by each model to the same CSV file as the averaged ensemble prediction for simple calculation of error metrics following execution

cmd_pred = (
    f'chemprop predict '
    f'--test-path reaxys_test_filtered_no_reg_nums.csv '
	  f'--model-paths '
    f'/content/drive/MyDrive/chemprop_runs/modern_chemprop_reaxys/replicate_0/model_0/best.pt '
    f'/content/drive/MyDrive/chemprop_runs/modern_chemprop_reaxys/replicate_1/model_0/best.pt '
    f'/content/drive/MyDrive/chemprop_runs/modern_chemprop_reaxys/replicate_2/model_0/best.pt '
    f'/content/drive/MyDrive/chemprop_runs/modern_chemprop_reaxys/replicate_3/model_0/best.pt '
    f'/content/drive/MyDrive/chemprop_runs/modern_chemprop_reaxys/replicate_4/model_0/best.pt '
    f'--smiles-columns Chromophore Solvent '
    f'--preds-path /content/drive/MyDrive/chemprop_runs/modern_chemprop_reaxys/modern_chemprop_reaxys_preds.csv'
    )
```

```python
# executing Chemprop prediction command with the Reaxys test subset
! /usr/local/miniconda/bin/conda run -n newchemprop \
bash -c "$cmd_pred"
```
