## **Creating Chemprop Training Command**

```python
# creating a command from important arguments/parameters within the 'verbose.log' and 'args.json' files attached by the authors, and printing it - 
# --gpu 0' ensures the utilised runtime is the GPU (A100) in this case
cmd = (
f'chemprop_train '
f'--data_path d4c_train.csv '
f'--dataset_type regression '
f'--num_folds 10 '
f'--epochs 200 '
f'--number_of_molecules 2 '
f'--depth 5 '
f'--hidden_size 1900 '
f'--ffn_hidden_size 1900 '
f'--ffn_num_layers 3 '
f'--dropout 0.1 '
f'--aggregation mean '
f'--batch_size 50 '
f'--metric rmse '
f'--extra_metrics mae '
f'--seed 0 '
f'--gpu 0 '
f'--save_dir /content/drive/MyDrive/chemprop_runs/table1_d4c_model'
        )

print(cmd)
```

```python
# executing Chemprop training command with the Deep4Chem training subset
! /usr/local/miniconda/bin/conda run \
-n chemprop15 \
bash -c '$cmd'
```
